```python
from datetime import datetime

def timestamp_correction(x):

    # чи нема пропуску
    if pd.isna(x):
        return pd.NaT
    
    x = str(x).strip()
    
    if len(x) == 19:        # для формату з секундами
        return datetime.strptime(x, "%m.%d.%Y %H:%M:%S")
    elif len(x) == 16:       # для формату без секундами
        return datetime.strptime(x, "%d.%m.%Y %H:%M")
    else:
        return pd.NaT
```

```python
from datetime import datetime

def timestamp_correction(x):

    # чи нема пропуску
    if pd.isna(x):
        return pd.NaT
    
    x = str(x).strip()

    formats = [
        "%m.%d.%Y %H:%M:%S",   # для формату з секундами
        "%d.%m.%Y %H:%M",      # без секунд
        "%m.%d.%Y %H:%M",      # m,d без секунд
    ]

    for format in formats:
        try:
            return datetime.strptime(x, format)
        except:
            continue

    return pd.NaT
```

```python
# застос до колонки
df["Timestamp_corrected"] = df["Timestamp"].apply(timestamp_correction)
```
