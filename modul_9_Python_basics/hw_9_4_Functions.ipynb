2. Напишіть функцію `calculation` яка приймає на вхід 4 аргументи типу int - `a`, `b`, `c`, `d` і вертає результат обчислення такої формули: $ (a + b - c) * d $.

  Викличте функцію на наступному наборі аргументів і виведіть результат на екран:
  ```
  a = 16
  ```

````python
def calculation(a, b, c, d):
    return (a + b - c) * d

a = 16
b = 4
c = 9
d = 3
result = calculation(a, b, c, d)
print(result)
````

4. Напишіть функцію `multipy_numbers`, яка приймає два числа як аргументи і повертає їх добуток.

  Викличіть функцію `multipy_numbers` на парах чисел
- 10 і 2
- 7 і 8
- 1024 і 999.
  
  Виведіть результат кожного виклику.
  При виконанні цього завдання з використанням циклу - отримаєте 2х балів за нього :)

````python
def multipy_numbers(a, b):
    return a * b

pairs = [(10, 2), (7, 8), (1024, 999)]

for a, b in pairs:
    result = multipy_numbers(a,b)
    print(result)
````

При цьому функція перед діленням перевіряє, чи другий аргумент не дорівнює 0. Якщо другий аргумент не 0, фукнція вертає результат ділення, інакше - замість виконання ділення, вертає повідомлення на екран "Не можу виконати ділення на 0".

    Якщо типи, надані у функцію, не `int` - функція виводить повідомлення "Не працюю з такими типами даних.".

  2. Запустіть функцію на наступних парах чисел:
- 12 і 6
- 20 і 0
- 112 і 10
- 99 і 0
- 11 i 1.1

  3. Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"

````python
def int_division(a: int, b: int):
    if not isinstance(a, int) or not isinstance(b, int):
        return "Не працюю з такими типами даних."
    elif b == 0:
        return "Не можу виконати ділення на 0"
    else:
        return a/b

pairs = [(12, 6), (20, 0), (112, 10), (99, 0), (11, 1.1)]

for a, b in pairs:
    result = int_division(a, b)
    print(f"Результат роботи функції: ", result)


"""
"Чому в деяких випадках функція повернула None?"
у мене не повертає None. тому можу лише фантазувати чому.
Бачу, якщо замінити return на print то отримаємо None

"""
````
