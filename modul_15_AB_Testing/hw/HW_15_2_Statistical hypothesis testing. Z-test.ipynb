1. Запишіть параметри задачі у змінні Python:
    - `sample_mean` - $\bar{x}$, середнє значення в вибірці
    - `population_mean` - $\mu_0$, середнє значення в популяції (тобто те, з яким ми порівнюємо середнє вибіркове значення)
    - `population_std` - $\sigma$, вибіркове стандартне відхилення, яке ми вважаємо, що є рівним ст. відх. популяції, адже маємо достатньо велику вибірку.
    - `sample_size` - $n$, розмір вибірки
    - `alpha` - $\alpha$ рівень значущості
    
    **Увага!** Для обчислення стандартного відхилення маємо скористатись функцією `np.std(your_dataframe, ddof=1)`. Чому так - розбираємо в лекції про t-test.

7. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$. Для прийняття рішення зробіть і порівняння z-статистики з критичним значенням, і проаналізуйте p-value.

```python
# ПАРАМЕТРИ ЗАДАЧІ

sample_mean = data.mean()
population_mean = 20
population_std = np.std(data, ddof=1)
sample_size = len(data)
alpha = 0.05

print(f"sample_mean = {sample_mean}")
print(f"population_mean = {population_mean}")
print(f"population_std = {population_std}")
print(f"sample_size = {sample_size}")
print(f"alpha = {alpha}")
```

ГІПОТЕЗИ

гіпотеза $H_0$: $\mu_0$ = 20

альтернативна $H_a$: $\mu$ > 20

тест: правосторонній

```python
# СТАНДАРТНА ПОМИЛКА SE 

# SE = σ / √n
standard_error = population_std / np.sqrt(sample_size)

print(f"формула: SE = σ / √n")
print(f"SE = {population_std} / √{sample_size}")
print(f"SE = {population_std} / {np.sqrt(sample_size):.3f}")
print(f"SE = {standard_error:.3f}")
print(f"похибка оцінки середнього за вибіркою становить приблизно ±{standard_error:.3f}")
```

```python
# Z-СТАТИСТИКА

# формула: z = (x̄ - μ₀) / SE
z_statistic = (sample_mean - population_mean) / standard_error

print(f"формула: z = (x̄ - μ₀) / SE")
print(f"z = ({sample_mean} - {population_mean}) / {standard_error:.3f}")
print(f"z = {sample_mean - population_mean} / {standard_error:.3f}")
print(f"z = {z_statistic:.3f}")

if z_statistic > 2:
    print("Досить велике відхилення")
elif z_statistic > 1:
    print("Помірне відхилення")
else:
    print("Невелике відхилення")
```

```python
# КРИТИЧНЕ z-значення
# для правостороннього тесту з α=0.05 передаємо (1 - alpha) = 0.95

z_critical = stats.norm.ppf(1 - alpha)

print(f"формула: z_critical = stats.norm.ppf(1 - alpha)")
print(f"z_critical = stats.norm.ppf(1 - {alpha})")
print(f"z_critical = {stats.norm.ppf(1 - alpha):.3f}")
```

```python
# P-VALUE
# p_value = 1 - stats.norm.cdf(z_statistic)
# правосторонній тест тому -> (1 - stats.norm.cdf(z))

p_value = 1 - stats.norm.cdf(z_statistic)
print(f"формула: p_value = 1 - stats.norm.cdf(z_statistic)")
print(f"p_value = {(1 - stats.norm.cdf(z_statistic)):.6f}")
print(f"{p_value:.4f} показує, наскільки отриманий результат узгоджується з H0 і як видно - ")
if p_value < 0.001:
    print("це НАДЗВИЧАЙНО малоймовірно! (менше 0.1%)")
elif p_value < 0.01:
    print("це дуже малоймовірно! (менше 1%)")
elif p_value < 0.05:
    print("це малоймовірно! (менше 5%)")
else:
    print("це цілком можливо випадково")
```

```python
# рішення, чи відхиляєте гіпотезу $H_0$

# 1: Порівняння z-статистики з критичним значенням
print(f"z-статистика = {z_statistic:.3f}")
print(f"критичне значення = {z_critical:.3f}")
print(f"порівняймо: {z_statistic:.3f} {'>' if z_statistic > z_critical else '≤'} {z_critical:.3f}")

if z_statistic > z_critical:
    print("Висновок: z-статистика перевищує критичне значення")
    decision = "відхилити H₀"
else:
    print("Висновок: z-статистика не перевищує критичне значення")
    decision = "не відхиляти H₀"

print(f"отже, наше рішення {decision}")
```

```python
# 2: Порівняння p-value з рівнем значущості (ЧАСТІШЕ ВИКОРИСТОВУЄТЬСЯ)
print(f"p_value = {p_value:.6f}")
print(f"рівень значущості = {alpha}")
print(f"порівняймо: {p_value:.6f} {'<' if p_value < alpha else '≥'} {alpha}")

if p_value < alpha:
    print("Висновок: p_value менше за рівень значущості")
    decision2 = "відхилити H₀"
else:
    print("Висновок: p_value не менше за рівень значущості")
    decision2 = "не відхиляти H₀"

print(f"отже, наше рішення {decision2}")
```

```python
# ВИСНОВОК

if decision2 == "відхилити H₀":
    print("Висновок: підтверджуємо, що нам вдалось статистично знучущо підвищити середній чек за липень 2025")
else:
    print("Висновок: нам не вдалось статистично знучущо підвищити середній чек за липень 2025")
```
