```python
df_city = df[["City", "gross income"]]
df_city
```

```python
filter_yangon = df['City'] == 'Yangon'
filter_mandalay = df['City'] == 'Mandalay'
```

```python
mean_yangon = df[filter_yangon]['gross income'].mean()
mean_mandalay = df[filter_mandalay]['gross income'].mean()

(float(mean_yangon), float(mean_mandalay))
```

```python
if mean_yangon > mean_mandalay:
    print("Cереднє значення gross income для міста Yangon > Mandalay")
else:
    print("Cереднє значення gross income для міста Yangon < Mandalay")
```

Unit price - ця колонка підсилить аналіз 2 попередніх гіпотез, оскільки покаже середню/мін/макс вартість товарів кожної категорії.
